```python
import math

# Input the radius(r)
r = float(input("Enter the radius of the sphere: "))

# Volume: (4/3) * pi * r^3
volume = (4/3) * math.pi * (r ** 3)

print(f"The volume of the sphere with radius {r} is: {volume:.2f}") 
```

```python
def calculate_area(side):
    return side * side

def calculate_perimeter(side):
    return 4 * side

def main():
    # Input side length(s)
    s = float(input("Enter the side length of the square: "))
    
    # Calculate area and perimeter 
    area = calculate_area(s)
    perimeter = calculate_perimeter(s)
    
    # Display the results
    print(f"Side length: {s}")
    print(f"Area of the square: {area:.2f}")
    print(f"Perimeter of the square: {perimeter:.2f}")

# Calling the main function
main()
```

```python
def check_case(character):
    if character.isupper():
        return "uppercase"
    elif character.islower():
        return "lowercase"
    else:
        return "not a letter"

def main():
    char = input("Enter a character: ")
    
    if len(char) != 1:
        print("Please enter a single character.")
        return
    
    result = check_case(char)
    
    print(f"The character '{char}' is {result}.")

# Call the main function
main()
```

```python
values = []

# Use a loop to get 5 values from the user
for i in range(5):
    num = float(input(f"Enter value {i+1}: "))
    values.append(num)

total = sum(values)

average = total / len(values)

print("\n--- Results ---")
print(f"Values entered: {values}")
print(f"Sum: {total}")
print(f"Average: {average:.2f}")
```
